#### Exercício 8

- chovendo = True
- guarda_chuva = False

```python
# Solução
chovendo = True
guarda_chuva = True
frase = ""

if chovendo == True:
    frase = "Está chovendo! "
else:
    frase = "Não está chovendo. "

if guarda_chuva == True:
    frase += "A pessoa em questão tem um guarda-chuva!"
else:
    frase += "A pessoa em questão não tem um guarda-chuva. :("

print(frase)
```

#### Exercício 29

Crie dois conjuntos: 

- set_a = {1, 2, 3, 4}
- set_b = {3, 4, 5, 6}

Encontre e imprima a interseção entre os dois conjuntos (os elementos que estão em ambos).

```python
# Solução
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

intersecao = set_a.intersection(set_b)

print(f"Interseção entre os dois conjuntos A e B: {intersecao}")
```

#### Exercício 30

Escreva um programa que peça ao usuário para digitar sua altura em metros (ex: 1.75) e seu peso em quilogramas (ex: 68.5). Calcule o Índice de Massa Corporal (IMC) usando a fórmula IMC = peso / (altura * altura) e imprima o resultado formatado com duas casas decimais.

```python
# Solução
altura_usuario = input("Favor informar a sua altura: ")
altura_usuario_int = float(altura_usuario)
peso_usuario = input("Favor informar o seu peso: ")
peso_usuario_int = int(peso_usuario)

imc = peso_usuario_int / (altura_usuario_int**2)

print(f"Seu IMC é: {imc:.2f}")
```
